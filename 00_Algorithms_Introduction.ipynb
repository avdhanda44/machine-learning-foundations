# Algorithms Introduction

This repository is organized like a Machine Learning course. Each topic has a theory notes notebook and a code companion notebook. The notes explain why the algorithm exists, how it thinks, the important formulas, and when to use it. The code companion then demonstrates the same mechanism in Python.


## How to Study This Repository

Study the notes first, then run the matching code notebook. The goal is not to memorize syntax. The goal is to understand the mechanism, then see that mechanism working in code.


## Main Families of Machine Learning

Supervised learning uses labeled data. Regression predicts numbers, while classification predicts categories. Unsupervised learning finds structure without labels. Ensemble learning combines multiple models to make stronger predictions. Dimensionality reduction compresses features while trying to preserve useful information.


## Course Map

The project starts with EDA, then regression, classification, tree-based models, ensembles, clustering, and PCA. The order is designed so each topic builds naturally on earlier ideas.
