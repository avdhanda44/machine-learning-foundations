# K-Means Code Companion

This notebook connects K-Means code with the theory: centroids, assignment, update, inertia, and choosing K.


```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## Create Unlabeled Data

K-Means is unsupervised, so the model receives features without target labels. Here we create sample points for clustering practice.


```python
X, true_groups = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=0.8,
    random_state=42
)

points = pd.DataFrame(X, columns=["feature_1", "feature_2"])
points.head()
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(points["feature_1"], points["feature_2"], alpha=0.75)
plt.xlabel("feature_1")
plt.ylabel("feature_2")
plt.title("Unlabeled Data")
plt.tight_layout()
plt.show()
```

## Fit K-Means

The model starts with centroids, assigns each point to the nearest centroid, moves centroids to the center of assigned points, and repeats until stable.


```python
kmeans = KMeans(n_clusters=4, n_init=10, random_state=42)
cluster_labels = kmeans.fit_predict(points)

points_with_clusters = points.copy()
points_with_clusters["cluster"] = cluster_labels
points_with_clusters.head()
```

## Inspect Centroids and Inertia

Centroids are the learned cluster centers. Inertia is the sum of squared distances from points to their assigned centroid.


```python
centroids = pd.DataFrame(kmeans.cluster_centers_, columns=["feature_1", "feature_2"])

print("Inertia:", kmeans.inertia_)
centroids
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(points["feature_1"], points["feature_2"], c=cluster_labels, cmap="viridis", alpha=0.75)
plt.scatter(centroids["feature_1"], centroids["feature_2"], c="red", marker="X", s=200, label="centroids")
plt.xlabel("feature_1")
plt.ylabel("feature_2")
plt.title("K-Means Clusters")
plt.legend()
plt.tight_layout()
plt.show()
```

## Choose K with Inertia and Silhouette Score

Inertia usually decreases as K increases. Silhouette score helps judge whether clusters are compact and separated.


```python
rows = []
for k in range(2, 9):
    model = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = model.fit_predict(points)
    rows.append({
        "k": k,
        "inertia": model.inertia_,
        "silhouette_score": silhouette_score(points, labels)
    })

k_results = pd.DataFrame(rows)
k_results
```

```python
plt.figure(figsize=(7, 4))
plt.plot(k_results["k"], k_results["inertia"], marker="o")
plt.xlabel("K")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.tight_layout()
plt.show()
```
