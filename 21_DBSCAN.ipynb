# DBSCAN Code Companion

This notebook demonstrates density-based clustering. The important idea is to see how `eps` and `min_samples` decide which points become core points, border points, and noise.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
```

## Create Non-Round Clusters

K-Means struggles with curved clusters because it prefers round groups. DBSCAN can handle this shape because it grows clusters through dense neighborhoods.


```python
X, _ = make_moons(n_samples=350, noise=0.07, random_state=42)
X[:5]
```

## Fit DBSCAN

`eps` is the neighborhood radius. `min_samples` is the minimum number of nearby points needed to make a dense region.


```python
dbscan = DBSCAN(eps=0.22, min_samples=5)
labels = dbscan.fit_predict(X)

pd.Series(labels).value_counts().sort_index()
```

## Visualize Clusters and Noise

DBSCAN labels noise as `-1`. A good setting should separate the main dense regions without treating too many normal points as noise.


```python
plt.figure(figsize=(7, 5))
for label in sorted(set(labels)):
    mask = labels == label
    name = 'Noise' if label == -1 else f'Cluster {label}'
    plt.scatter(X[mask, 0], X[mask, 1], s=35, label=name)
plt.title('DBSCAN Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

## Evaluate the Clustering

Silhouette score should be calculated only on points that belong to clusters, because noise points do not represent a normal cluster assignment.


```python
cluster_mask = labels != -1
if len(set(labels[cluster_mask])) > 1:
    score = silhouette_score(X[cluster_mask], labels[cluster_mask])
    print(f'Silhouette score without noise: {score:.3f}')
else:
    print('Not enough clusters to calculate silhouette score.')
```
