# Hierarchical Clustering Code Companion

This notebook shows how hierarchical clustering builds clusters by repeatedly merging nearby groups. The dendrogram helps visualize the merge history.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import linkage, dendrogram
```

## Prepare the Data

Scaling is helpful because distance-based clustering should not let one large-scale feature dominate the merge decisions.


```python
iris = load_iris(as_frame=True)
X = iris.data
X_scaled = StandardScaler().fit_transform(X)
X.head()
```

## Draw a Dendrogram

The dendrogram shows which points or clusters merge first. Shorter vertical merge lines mean groups were closer together.


```python
linked = linkage(X_scaled, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(linked, truncate_mode='level', p=4)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample index or merged cluster')
plt.ylabel('Distance')
plt.show()
```

## Fit Agglomerative Clustering

Here we choose three clusters because the Iris dataset is known to contain three flower species. In real work, the dendrogram and domain knowledge help choose this value.


```python
model = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = model.fit_predict(X_scaled)

pd.Series(labels).value_counts().sort_index()
```

## Evaluate Cluster Separation

Silhouette score checks whether samples are closer to their own cluster than to other clusters.


```python
score = silhouette_score(X_scaled, labels)
print(f'Silhouette score: {score:.3f}')
```
