# Gaussian Mixture Models Code Companion

This notebook demonstrates soft clustering. Instead of forcing each point into a cluster with complete certainty, a Gaussian Mixture Model estimates the probability that a point belongs to each component.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
```

## Create Overlapping Groups

Gaussian Mixture Models are useful when clusters overlap and we want membership probabilities instead of only hard labels.


```python
X, _ = make_blobs(n_samples=400, centers=3, cluster_std=[1.0, 1.7, 0.8], random_state=42)
X[:5]
```

## Fit a Gaussian Mixture Model

`n_components` tells the model how many Gaussian distributions to fit. Each component has its own mean, covariance, and weight.


```python
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(X)

labels = gmm.predict(X)
probabilities = gmm.predict_proba(X)

pd.DataFrame(probabilities, columns=['Component 0', 'Component 1', 'Component 2']).head()
```

## Visualize Hard Labels

The model can still give a single cluster label by choosing the component with the highest probability.


```python
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, s=35, cmap='viridis')
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c='red', s=150, marker='x', label='Component means')
plt.title('Gaussian Mixture Model Clusters')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

## Inspect Model Quality

AIC and BIC compare mixture models while penalizing unnecessary complexity. Lower values are usually preferred.


```python
print(f'Silhouette score: {silhouette_score(X, labels):.3f}')
print(f'AIC: {gmm.aic(X):.1f}')
print(f'BIC: {gmm.bic(X):.1f}')
```
