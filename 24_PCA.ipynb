# PCA Code Companion

This notebook shows how PCA reduces many features into fewer principal components while preserving as much variance as possible.


```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Prepare and Scale Data

PCA is sensitive to scale because it is based on variance. Standardizing the features keeps large-scale columns from dominating the components.


```python
data = load_breast_cancer(as_frame=True)
X = data.data
y = data.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X.shape
```

## Reduce to Two Components

The first component captures the largest possible variance. The second captures the largest remaining variance while staying perpendicular to the first.


```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
pca_df['target'] = y
pca_df.head()
```

## Explained Variance

Explained variance ratio tells us how much information each principal component preserves from the original data.


```python
for i, ratio in enumerate(pca.explained_variance_ratio_, start=1):
    print(f'PC{i}: {ratio:.3f}')
print(f'Total preserved variance: {pca.explained_variance_ratio_.sum():.3f}')
```

## Visualize the Compressed Data

The plot uses only two PCA components, but each point still represents information from all original features.


```python
plt.figure(figsize=(7, 5))
for target_value, label in enumerate(data.target_names):
    mask = pca_df['target'] == target_value
    plt.scatter(pca_df.loc[mask, 'PC1'], pca_df.loc[mask, 'PC2'], s=30, label=label)
plt.title('Breast Cancer Data Projected with PCA')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```

## Choose Number of Components

In practice, we often choose enough components to preserve a target amount of variance, such as 95%.


```python
pca_95 = PCA(n_components=0.95, random_state=42)
pca_95.fit(X_scaled)
print(f'Components needed for 95% variance: {pca_95.n_components_}')
```
